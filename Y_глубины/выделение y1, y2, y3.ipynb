```python
import pandas as pd

# Загрузка данных из исходного файла
df_real = pd.read_csv('C:/Users/fusb1/Desktop/Для диплома/Baikal 1 Temp Secchi Chlf raw data final.csv')

# Преобразуем столбец "Дата" в формат даты и оставляем только дату
df_real['date'] = pd.to_datetime(df_real['Дата'], errors='coerce').dt.date

# Фильтруем только те строки, где дата после 1995 года
df_real = df_real[df_real['date'] >= pd.to_datetime('1995-01-01').date()]

# Создаём основной датафрейм
df_main = df_real[['date', 'Chl_a', 'Горизонт']].copy()
df_main.columns = ['date', 'хлорофилл', 'горизонт']

# Сохраняем основной файл
df_main.to_csv('C:/Users/fusb1/Desktop/После Рыбки/Y_глубины/Байкал_глубина_all.csv', index=False)

# Создаём и сохраняем датафреймы для разных горизонтов
def prepare_depth_df(df, depth):
    df_depth = df[df['горизонт'] == depth].copy()
    return df_depth.drop('горизонт', axis=1)

# Для горизонта 0
df_surface = prepare_depth_df(df_main, 0)
df_surface.to_csv('C:/Users/fusb1/Desktop/После Рыбки/Y_глубины/Y1_(глубина_=_0).csv', index=False)

# Для горизонта 5
df_10 = prepare_depth_df(df_main, 5)
df_10.to_csv('C:/Users/fusb1/Desktop/После Рыбки/Y_глубины/Y2_(глубина_=_5).csv', index=False)

# Для горизонта 10
df_20 = prepare_depth_df(df_main, 10)
df_20.to_csv('C:/Users/fusb1/Desktop/После Рыбки/Y_глубины/Y3_(глубина_=_10).csv', index=False)

# Выводим первые 5 строк каждого для проверки
print("Основной файл (все горизонты):")
print(df_main.head())

print("\nПоверхность (горизонт = 0):")
print(df_surface.head())

print("\nГлубина 5м:")
print(df_10.head())

print("\nГлубина 10м:")
print(df_20.head())
```
